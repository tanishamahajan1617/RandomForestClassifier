```python
file_path = "/kaggle/input/iris/Iris.csv"

df = pd.read_csv(file_path)
df
```

```python
df.drop(columns="Id")


x = df[['SepalLengthCm','SepalWidthCm','PetalLengthCm','PetalWidthCm']]
y = df['Species']


```

```python
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import LabelEncoder,StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.tree import plot_tree
y_encoded = LabelEncoder().fit_transform(y)

x_train , x_test , y_train, y_test = train_test_split(x,y_encoded, test_size = 0.2, random_state = 42)

params = {
    'RF__n_estimators':[50,20,10,100],
    'RF__max_features':[12,10,70,30],
    'RF__max_depth':[10,20,5,6],
    'RF__max_samples':[10,20,9,8]
}



pipe_rf = Pipeline([('standard_scaler',StandardScaler()),
                   ('RF',RandomForestClassifier(random_state=42))])



```

```python
grid = GridSearchCV(pipe_rf, params,cv=5,n_jobs=-1)
grid.fit(x_train,y_train)
```

```python
best_rf= grid.best_estimator_
print(grid.best_score_)
```

```python
import joblib
from sklearn.metrics import accuracy_score
joblib.dump(grid.best_estimator_, 'randomclassifier_pipeline.pkl')
print("Pipeline saved! Test Accuracy:", accuracy_score(y_test, grid.best_estimator_.predict(x_test)))
```
